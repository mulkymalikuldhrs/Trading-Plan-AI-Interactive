## Step 3: Configure Your API URL

Replace the placeholder URL below with the actual Web App URL you got from deploying your Google Apps Script.

```python
from dhaher_ai_client import DhaherAiClient
import pandas as pd
import json

# IMPORTANT: Replace this with your actual URL
APPS_SCRIPT_URL = "YOUR_GOOGLE_APPS_SCRIPT_WEB_APP_URL"

try:
    client = DhaherAiClient(APPS_SCRIPT_URL)
    print("✅ Client initialized successfully!")
except ValueError as e:
    print(f"❌ Error: {e}")
    print("Please update the APPS_SCRIPT_URL variable with your deployed Google Apps Script URL.")
```

## Step 4: Fetch and Analyze Your Trading Journal

Let's get your entire trading history and analyze it using pandas.

```python
try:
    journal_data = client.get_journal_data()
    df = pd.DataFrame(journal_data)
    
    print(f"Successfully fetched {len(df)} journal entries.")
    
    # Display the last 5 trades
    display(df.tail())
    
    # Basic Analysis
    print("\n--- Basic Analysis ---")
    win_rate = (df[df['Result'] == 'WIN'].shape[0] / df.shape[0]) * 100
    print(f"Win Rate: {win_rate:.2f}%")
    
    average_pnl = df['PnL'].astype(float).mean()
    print(f"Average PnL: ${average_pnl:.2f}")
    
except Exception as e:
    print(f"❌ An error occurred: {e}")
```

## Step 5: Get an On-Demand AI Forecast

Let's ask the AI for its latest forecast on GOLD.

```python
try:
    forecast = client.get_forecast('GOLD')
    
    print("🔮 AI Forecast for GOLD:")
    print(json.dumps(forecast, indent=2))
    
except Exception as e:
    print(f"❌ An error occurred: {e}")
```

## Step 6: Log a New Trade Programmatically

```python
new_trade = {
    'pair': 'BTC/USD',
    'direction': 'Buy',
    'entry': 62000,
    'sl': 61000,
    'tp': 67000,
    'rrr': 5,
    'setup': 'Colab Test',
    'mood': 'Analytical',
    'ai_status': 'Approved',
    'result': 'WIN',
    'emotion_after': 'Satisfied',
    'gpt_comment': 'Logged programmatically from Google Colab.'
}

try:
    response = client.log_trade(new_trade)
    print(f"✅ {response}")
except Exception as e:
    print(f"❌ An error occurred: {e}")
```
